```python
def print_board(board):
  for row in board:
    print("  ".join(row)) # Fix: Removed '.' and added join

def is_safe(board,row,col):
  for i in range(row):
    if board[i][col]=='Q':
      return False

  # Fix: Corrected indentation and logic for diagonal checks
  i_diag = row - 1
  j_diag = col - 1
  while i_diag >= 0 and j_diag >= 0:
    if board[i_diag][j_diag]=='Q':
      return False
    i_diag -= 1
    j_diag -= 1

  i_diag = row - 1
  j_diag = col + 1
  while i_diag >= 0 and j_diag < 8:
    if board[i_diag][j_diag]=='Q':
      return False
    i_diag -= 1
    j_diag += 1

  return True

def solve(board,row):
  if row==8:
    return True

  for col in range(8):
    if is_safe(board,row,col):
      board[row][col]='Q' # Fix: Changed '==' to '='

      if solve(board,row+1):
        return True

      board[row][col]='.'

  return False


board=[['.' for _ in range(8)] for _ in range(8)] # Fix: Corrected 'for_in' to 'for _ in'

if solve(board,0):
  print("\nSolution for 8-Queens Problem:\n")
  print_board(board)
else:
  print("No Solution Exists")
```
