```python
from collections import deque

def bfs(graph, start):
    visited = set()
    queue = deque([start])
    visited.add(start)

    print("BFS Traversal: ", end="")

    while queue:
        vertex = queue.popleft()
        print(vertex, end=" ")

        # Corrected: Indented this block to be inside the while loop
        if vertex in graph:
            for neighbor in graph[vertex]:
                if neighbor not in visited:
                    visited.add(neighbor)
                    queue.append(neighbor)
    print()

# --- Corrected logic: Get input first, then run BFS ---
graph = {}
n = int(input("ENTER THE NUMBER OF VERTICES: "))

for i in range(n):
    vertex_name = input(f"ENTER THE VERTEX {i+1}: ")
    neighbors = input(f"ENTER THE NEIGHBORS OF {vertex_name} (space-separated): ").split()
    graph[vertex_name] = neighbors

start_node = input("ENTER THE STARTING VERTEX: ")
print("\nGraph:", graph)

if start_node in graph:
    bfs(graph, start_node)
else:
    print(f"Error: Start vertex '{start_node}' not found in the graph.")
```
