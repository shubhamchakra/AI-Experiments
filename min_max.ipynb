```python
def minimax(depth, nodeIndex, isMax, Scores, Height):

    if depth == Height:
        return Scores[nodeIndex]

    if isMax:
        return max (
           minimax(depth + 1, nodeIndex * 2,
                          False, Scores, Height),
            minimax(depth + 1, nodeIndex * 2 + 1,
                   False, Scores, Height))

    else:
       return min (
        minimax(depth + 1, nodeIndex * 2,
               True, Scores, Height),
        minimax(depth + 1, nodeIndex * 2 + 1,
               True, Scores, Height))


Scores = list(map(int, input("Enter leaf node values : ").split()))
Height = 3
result =minimax(0,0,True,Scores,Height)
print("The Optimal Solutio is ",result)
```
