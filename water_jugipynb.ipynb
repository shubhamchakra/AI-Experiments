```python
from collections import deque

def water_jug(capacity1, capacity2, target):
  visited = set()
  queue = deque()

  # Initial state: (jug1_amount, jug2_amount, current_path)
  queue.append((0, 0, []))

  while queue:
    jug1, jug2, path = queue.popleft()

    # If this state has been visited, skip to avoid redundant processing
    if (jug1, jug2) in visited:
      continue

    visited.add((jug1, jug2))

    # Add current state to path
    current_path = path + [(jug1, jug2)]

    # Check if target is reached
    if jug1 == target or jug2 == target:
      return current_path

    # Generate next possible states
    possible_next_states = []

    # 1. Fill Jug 1
    possible_next_states.append((capacity1, jug2))
    # 2. Fill Jug 2
    possible_next_states.append((jug1, capacity2))
    # 3. Empty Jug 1
    possible_next_states.append((0, jug2))
    # 4. Empty Jug 2
    possible_next_states.append((jug1, 0))

    # 5. Pour Jug 1 to Jug 2
    amount_to_pour = min(jug1, capacity2 - jug2)
    possible_next_states.append((jug1 - amount_to_pour, jug2 + amount_to_pour))

    # 6. Pour Jug 2 to Jug 1
    amount_to_pour = min(jug2, capacity1 - jug1)
    possible_next_states.append((jug1 + amount_to_pour, jug2 - amount_to_pour))

    for next_jug1, next_jug2 in possible_next_states:
      if (next_jug1, next_jug2) not in visited:
        queue.append((next_jug1, next_jug2, current_path))

  return None # No solution found after exploring all states

# --- Example Usage (moved outside the function) ---

jug1_capacity = int(input("Enter the capacity of jug 1:"))
jug2_capacity = int(input("Enter the capacity of jug 2:"))
target_volume = int(input("Enter the target volume:"))

solution = water_jug(jug1_capacity, jug2_capacity, target_volume)

if solution:
  print("\nSteps to reach the target:")
  for state in solution:
    print(f"Jug 1: {state[0]}, Jug 2: {state[1]}")
else:
  print("\nNo solution found.")
```
